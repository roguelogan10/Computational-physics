### Creation d'objet
L'utilisation de vpython passe essentiellement par la definition des objets qu'on veut utiliser et de la mise à jour de leurs caractéristiques. Vpython s'occupe du reste, c'est à dire la creation des objets, leurs ombres, leurs animations...
 Il existe un grand nombre d'objet 3D dans la librairie vpython, par exemple la sphère, le parallelépipède, le cylindre, l'ellipse, la fléche,le cone ...

 Chaque objet posséde des attributs qui leurs sont propres mais les plus communs sont la position dans l'espace (x,y,z), et la coleurs

##### Exemple d'attribut d'une sphère
- Rayon ( raduis)
  
- Couleur (color)
  
- Position (pos)

##### Exemple d'attribut d'un cone

- color 

- pos
  
- axis
 
- radius
